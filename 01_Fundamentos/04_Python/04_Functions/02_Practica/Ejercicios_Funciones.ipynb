## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string.

````python

frase= "Tres tristes tigres trigaban trigo en un trigal"

def ejer_5(frase):
    diccionario_contador = {}
    frase = frase.replace(" ","")
    frase = frase.lower()
    
    for letra in frase:
        diccionario_contador[letra]= diccionario_contador.get(letra, 0) + 1 
    return diccionario_contador

ejer_5(frase)
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

````python
lista_6 = [1,2,3,4,5,6]
copia_6 = lista_6.copy()

def ejer_6(lista_6, cmnd, elemento):
    if cmnd == "add":
        copia_6.append(elemento)
    elif cmnd == "remove":
        if elemento in lista_6:
            copia_6.remove(elemento)
    return lista_6

# Comprobación: 

ejer_6(copia_6, "remove", 6)

````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python

def fibonacci (x):
    if x == 0:
        return 0
    elif x == 1:
        return 1
    else:
        return fibonacci (x -1) + fibonacci (x - 2)


x = int(input("Dime un número"))

print("El número" ,x, "es", fibonacci(x), "en la sucesión de fibonacci")
````
