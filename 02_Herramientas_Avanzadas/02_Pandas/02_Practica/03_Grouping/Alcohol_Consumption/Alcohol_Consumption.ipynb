```python
# import pandas
import pandas as pd

class display(object):
    """Display HTML representation of multiple objects"""
    
    template = """<div style="float: left; padding: 10px;">
    <p style='font-family:"Courier New", Courier, monospace'>{0}</p>{1}
    </div>"""
    def __init__(self, *args):
        self.args = args
        
    def _repr_html_(self):
        return '\n'.join(self.template.format(a, eval(a)._repr_html_())
                         for a in self.args)
    
    def __repr__(self):
        return '\n\n'.join(a + '\n' + repr(eval(a))
                           for a in self.args)
```

### Step 4. Which continent drinks more beer on average?

```python
beer_per_continent = drinks.groupby('continent')['beer_servings'].mean().round(2)
beer_per_continent
```

Europa

```python
print(beer_per_continent.loc['EU'])
```

### Step 5. For each continent print the statistics for wine consumption.

```python
continent_wine_stats = drinks.groupby('continent')['wine_servings'].describe().round(2)
continent_wine_stats
```

### Step 6. Print the mean alcohol consumption per continent for every column

```python
mean_alcohol_continent = drinks.groupby("continent").mean(numeric_only= True).round(2)
mean_alcohol_continent

```

### Step 7. Print the median alcohol consumption per continent for every column

```python
median_alcohol_continent  = drinks.groupby('continent').median(numeric_only=True).round(2)
median_alcohol_continent
```

### Step 8. Print the mean, min and max values for spirit consumption.
#### This time output a DataFrame

```python
min_max_mean_spirit = drinks.groupby('continent')['spirit_servings'].agg(['mean','min','max']).round(2)
min_max_mean_spirit
```
