```python
import pandas as pd 
import random as random 


class display(object):
    """Display HTML representation of multiple objects"""
    
    template = """<div style="float: left; padding: 10px;">
    <p style='font-family:"Courier New", Courier, monospace'>{0}</p>{1}
    </div>"""
    def __init__(self, *args):
        self.args = args
        
    def _repr_html_(self):
        return '\n'.join(self.template.format(a, eval(a)._repr_html_())
                         for a in self.args)
    
    def __repr__(self):
        return '\n\n'.join(a + '\n' + repr(eval(a))
                           for a in self.args)
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user). 

### Step 3. Assign it to a variable called users.

```python
users = pd.read_csv("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user", sep = "|")
```

```python
users.set_index("user_id", inplace= True)
```

### Step 4. Discover what is the mean age per occupation

```python
users.groupby('occupation')['age'].mean().round(1)

#el transform es el groupby desenrollado
#sin las mismas dimensiones no se pueen juntar
```

### Step 5. Discover the Male ratio per occupation and sort it from the most to the least

```python
male_ratio = users.groupby(['occupation','gender'])['age'].count() 
```

```python

male_ratio = users.groupby('occupation')['gender'].apply(lambda x: (x == 'M').sum() / x.count())

male_ratio_sorted = male_ratio.sort_values(ascending=False)

print(male_ratio_sorted)
```

### Step 6. For each occupation, calculate the minimum and maximum ages

```python
age_ranges = users.groupby(['occupation'])['age'].agg(
    min_age ='min',
    max_age = 'max'
    ) .sort_values(by= 'min_age')

age_ranges
```

### Step 7. For each combination of occupation and gender, calculate the mean age

```python
occupation_and_gender_mean_age = users.groupby(['occupation', 'gender'])['age'].mean().round(1)

occupation_and_gender_mean_age
```

### Step 8.  For each occupation present the percentage of women and men

```python

occ_percent = users.groupby('occupation')['gender'].value_counts(normalize=True) * 100
occ_percent = occ_percent.unstack().fillna(0).round(2)

occ_percent
```
