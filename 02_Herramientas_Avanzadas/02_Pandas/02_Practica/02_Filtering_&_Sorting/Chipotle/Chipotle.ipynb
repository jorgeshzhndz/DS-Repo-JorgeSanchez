### Step 1. Import the necessary libraries

```python
import pandas as pd
import random as random

class display(object):
    """Display HTML representation of multiple objects"""
    
    template = """<div style="float: left; padding: 10px;">
    <p style='font-family:"Courier New", Courier, monospace'>{0}</p>{1}
    </div>"""
    def __init__(self, *args):
        self.args = args
        
    def _repr_html_(self):
        return '\n'.join(self.template.format(a, eval(a)._repr_html_())
                         for a in self.args)
    
    def __repr__(self):
        return '\n\n'.join(a + '\n' + repr(eval(a))
                           for a in self.args)
```

# Name of the max valued product

```python
#Hay q cambiar los precos a numeros
chipo['item_price'] = chipo['item_price'].str.replace('$', '').astype(float)
```

```python
max_val_prod = chipo.loc[chipo['item_price'].idxmax(), 'item_name']
precio_max = chipo.item_price.max()
print(f"El producto más caro es {max_val_prod} y vale {precio_max}$")
```

### Step 4. How many products cost more than $10.00?

```python
print(chipo[chipo.item_price > 10].item_name.count())
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
canned_soda_orders = chipo[(chipo['item_name'] == 'Canned Soda') 
                  & (chipo['quantity'] > 1)]

print(len(canned_soda_orders))
```

```python
# Filter for item_name 'Canned Soda' and quantity greater than 1
canned_soda_orders = chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1)]

# Get the number of times this occurred
print(len(canned_soda_orders))
# Output: 20
```
