## Ejercicio 8
Concatena los siguientes arrays verticamente

```Python
a = np.arange(10).reshape(2,-1)
b = np.repeat(1, 10).reshape(2,-1)
```

````python
a = np.arange(10).reshape(2,-1)
print(a,"\n")

b = np.repeat(1, 10).reshape(2,-1)
print(b, "\n")
````

````python
np.concatenate([a,b], axis = 0) #concatenate solo acepta un argumento
````
