## Ejercicio 8
Dado el siguiente array:

```Python

```

Filtra el array para quedarte únicamente con los loros.

Resuélvelo de dos maneras diferentes

````python
x = np.array(["Loro", "Perro", "Gato", "Loro", "Perro"])
````

````python


# Método 1:  Bool (Mask)
mask = x == "Loro"
resultado_1 = x[mask]
print("1:", resultado_1)

# Método 2:  np.where()
where_2 = np.where(x == "Loro")
resultado_2 = x[where_2]
print("2:", resultado_2)
````

## Ejercicio 19
Crea una matriz aleatoria de 5x5 y estandariza sus valores. Para normalizar tienes que restar sus valores por la media(`np.mean`) y dividir el resultado por la desviación estándar(`np.std`).

````python
ejer_19 = np.round(np.random.rand(5,5), 3)

mean_19 = np.round(np.mean(ejer_19),3)
print("La media es:", mean_19)
std_19 = np.round(np.std(ejer_19), 3)
print("La desviacion estandar es:", std_19)

norm_19 = np.round((ejer_19-mean_19)/std_19, 3)

print("Normalización:", norm_19)
````
